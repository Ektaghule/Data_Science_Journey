Evaluate the measures of central tendency for the whole data followed by the simple random sample, stratified sample, cluster sample and systematic sample.
Compare the means of all the different types of sample.
Finally, conclude which specific technique mean value is almost similar to the population mean value.

```python
population_mean = data['Income'].mean()
print("Population")
print(f"Mean: {population_mean}")
population_median = data['Income'].median()
print(f"Median: {population_median}")
population_mode = data['Income'].mode()[0]
print(f"Mode: {population_mode}")
```

```python
sample_size = int(len(data)*0.20)
```

```python
population_simple_random_sample = data.sample(n=sample_size, random_state=42)
print("Simple Random Sampling")
random_mean = population_simple_random_sample['Income'].mean()
print(f"Mean: {random_mean}")
random_median = population_simple_random_sample['Income'].median()
print(f"Median: {random_median}")
random_mode = population_simple_random_sample['Income'].mode()[0]
print(f"Mode: {random_mode}")
```

```python
population_stratified_sample = data.groupby("Education").apply(lambda x: x.sample(min(len(x),20), random_state=42))

print("Stratified Sampling")
stratified_mean = population_stratified_sample['Income'].mean()
print(f"Mean: {stratified_mean}")
stratified_median = population_stratified_sample['Income'].median()
print(f"Median: {stratified_median}")
stratified_mode = population_stratified_sample['Income'].mode()[0]
print(f"Mode: {stratified_mode}")
```

```python
population_clusters = data['Education'].unique()


population_selected_clusters = np.random.choice(population_clusters, size=3, replace=False)


population_clustered_sample = data[data['Education'].isin(population_selected_clusters)]


print("Cluster Sampling")
clustered_mean = population_clustered_sample['Income'].mean()
print(f"Mean: {clustered_mean}")
clustered_median = population_clustered_sample['Income'].median()
print(f"Median: {clustered_median}")
clustered_mode = population_clustered_sample['Income'].mode()[0]
print(f"Mode: {clustered_mode}")
```

```python
population_systematic_sample = data.iloc[::10, :]

print("Systematic Sampling")
systematic_mean = population_systematic_sample['Income'].mean()
print(f"Mean: {systematic_mean}")
systematic_median = population_systematic_sample['Income'].median()
print(f"Median: {systematic_median}")
systematic_mode = population_systematic_sample['Income'].mode()[0]
print(f"Mode: {systematic_mode}")
```

```python
results = {
    "Population":{
        "Mean": population_mean,
        "Median": population_median,
        "Mode": population_mode
    },
    "Random Sampling":{
        "Mean": random_mean,
        "Median": random_median,
        "Mode": random_mode
    },
    "Stratified Sampling": {
        "Mean": stratified_mean,
        "Median": stratified_median,
        "Mode": stratified_mode
    },
    "Cluster Sampling":{
        "Mean": clustered_mean,
        "Median": clustered_median,
        "Mode": clustered_mode
    },
    "Systematic Sampling":{
        "Mean": systematic_mean,
        "Median": systematic_median,
        "Mode": systematic_mode
    }
}

pd.DataFrame(results).T
```
