Data creation of the nominal column named 'colors'

```python
import numpy as np
import pandas as pd

np.random.seed(0)
num_samples = 1000

colors = ['Red', 'Blue', 'Green', 'Yellow', 'Purple', 'Orange']
nominal_data = np.random.choice(colors, num_samples)
nominal_data
```

```python
dataframe = pd.DataFrame(nominal_data, columns=['Colors'])
dataframe
```

Creation of the ordinal column named 'satisfaction_levels'

```python
np.random.seed(0)

satisfaction_levels = ['Low', 'Medium', 'High']
ordinal_data = np.random.choice(satisfaction_levels, num_samples)

dataframe['Satisfaction levels'] = ordinal_data
dataframe.head()
```

Data creation of the interval data under Temperature column

```python
np.random.seed(0)
interval_data = np.random.randint(0, 100, num_samples)
dataframe['Temperature'] = interval_data
dataframe
```

Data creation of the range data under the income column

```python
np.random.seed(0)
ratio_data = np.random.randint(10000, 1000000, num_samples)
dataframe['Income'] = ratio_data
dataframe
```

Save the file in csv

```python
dataframe.to_csv('Types_of_data.csv')
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

#Plot of ratio data
plt.hist(dataframe['Income'], bins=20, color='green', alpha=0.7, edgecolor='black')
plt.title('Income Distribution')
plt.xlabel('Income(Rupees)')
plt.ylabel('Frequency')
plt.show()
```
